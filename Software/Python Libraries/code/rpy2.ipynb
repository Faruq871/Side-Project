# rpy2

Intermediate reference for evaluating R code and converting data between Python and R. A compatible R installation is required.

## Access R

```python
import rpy2.robjects as ro

r = ro.r
r('R.version.string')
```

`rpy2.robjects` exposes R objects and evaluation from Python. `ro.r` is the embedded R interface. A compatible R installation and environment configuration are required; check compatibility among Python, R, and rpy2 versions.

## Evaluate R expressions

```python
result = r('x <- c(1, 2, 3); mean(x)')
list(result)
```

Calling `r()` evaluates an R expression and returns an R object. Convert simple vectors to Python lists with `list()`. For larger workflows, define R functions and call them instead of building long expression strings.

## Call R functions

```python
mean = r['mean']
mean(ro.FloatVector([2,4,6]))
```

Use `r['name']` to retrieve an R function or object. `FloatVector`, `IntVector`, and `StrVector` construct R vectors from Python sequences. Returned values are R objects and may need explicit conversion.

## Convert pandas and R data frames

```python
import pandas as pd
from rpy2.robjects import pandas2ri, default_converter
from rpy2.robjects.conversion import localconverter

pdf = pd.DataFrame({'x':[1,2], 'y':[3.5,4.5]})
with localconverter(default_converter + pandas2ri.converter):
    r_df = ro.conversion.py2rpy(pdf)
    back = ro.conversion.rpy2py(r_df)
```

`pandas2ri.converter` defines conversions between pandas and R data frames. `localconverter()` activates conversion only inside its context. Explicit conversion avoids assuming that every R object has a direct Python equivalent.

## R packages

```python
r('library(stats)')
r('summary(c(2, 4, 6))')
```

Load installed R packages with `library()`. Package installation is usually managed by R's `install.packages()` and may require repository access and system dependencies.
