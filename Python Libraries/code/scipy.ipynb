# SciPy

Intermediate reference for scientific routines built on NumPy arrays.

## Optimization

```python
import numpy as np
from scipy.optimize import minimize, root

result = minimize(lambda x: (x[0]-3)**2, x0=[0.])
result.x, result.fun
```

`minimize()` finds a local minimum of a scalar objective. Supply an initial value through `x0`; bounds, constraints, gradients, and algorithms can also be specified. `root()` solves systems of nonlinear equations. Inspect result success flags and messages.

## Statistics and tests

```python
from scipy import stats

stats.norm.cdf(1.96)
stats.ttest_ind([2,3,4], [5,6,7], equal_var=False)
stats.pearsonr([1,2,3], [2,4,5])
```

`scipy.stats` contains distributions, descriptive statistics, and hypothesis tests. Distribution objects expose `pdf()`, `cdf()`, `ppf()`, and `rvs()`. `ttest_ind()` compares independent samples; `pearsonr()` returns a correlation and p-value. Check assumptions and report effect sizes.

## Linear algebra

```python
from scipy import linalg

A = np.array([[3.,1.],[1.,2.]])
linalg.solve(A, [9.,8.])
linalg.svd(A), linalg.det(A)
```

`scipy.linalg` provides dense linear algebra routines including `solve()`, `inv()`, `det()`, `svd()`, `eig()`, `qr()`, and `cholesky()`. It overlaps with `numpy.linalg` and includes additional routines.

## Sparse matrices

```python
from scipy import sparse

S = sparse.csr_matrix([[0,1,0],[2,0,3]])
S.nnz, S.toarray(), S @ np.array([1,2,3])
```

Sparse formats store nonzero values efficiently. CSR is useful for arithmetic and row slicing; CSC is useful for column-oriented operations; COO is convenient for construction. `nnz` reports stored nonzero entries. Avoid converting large sparse matrices to dense arrays.

## Interpolation and signals

```python
from scipy.interpolate import interp1d
from scipy import signal

f = interp1d([0,1,2], [0,2,1], kind='linear')
f([.5,1.5])
b, a = signal.butter(3, .2)
```

`interp1d()` builds an interpolation function; alternatives include `CubicSpline` and `PchipInterpolator`. `signal` provides filtering, convolution, spectral analysis, and peak detection. `butter()` designs a Butterworth filter; `filtfilt()` applies a zero-phase forward/backward filter.
