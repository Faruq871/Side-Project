# pythonnet

Intermediate reference for loading .NET assemblies and using .NET classes from Python. A compatible .NET runtime is required.

## Initialize runtime

```python
from pythonnet import load

load('coreclr')  # Initialize before importing clr
import clr
```

`load()` selects and initializes a .NET runtime. The runtime must be installed and compatible with the Python process. Runtime selection must happen before importing `clr` or loading assemblies.

## Load assemblies and types

```python
clr.AddReference('System')
from System import String, DateTime

String('Hello from .NET'), DateTime.Now
```

`AddReference()` loads a .NET assembly by name or path. Namespaces and types can then be imported as Python modules. Use an explicit assembly path when the assembly is not discoverable.

## Use .NET collections

```python
from System.Collections.Generic import List

items = List[String]()
items.Add('alpha'); items.Add('beta')
items.Count, items[0]
```

Generic .NET collections use Python.NET's type syntax. `List[String]()` creates a strongly typed list. Methods and properties are exposed through Python attribute access; overload resolution follows .NET rules.

## Call .NET methods

```python
from System import Math

Math.Sqrt(81), Math.Round(3.14159, 2)
```

Static .NET methods can be called through imported types. Primitive values are often converted automatically, but overloaded methods may require explicit .NET types. Consult the assembly API for method signatures.

## Handle exceptions

```python
from System import InvalidOperationException

try:
    raise InvalidOperationException('Example failure')
except InvalidOperationException as exc:
    print(exc)
```

Catch .NET exceptions using their imported exception types. Keep exception boundaries clear when calling methods that may throw .NET-specific errors.
