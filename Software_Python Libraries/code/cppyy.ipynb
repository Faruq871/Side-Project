# cppyy

Intermediate reference for exposing C++ declarations and classes to Python. A supported C++ compiler is required.

## Declare C++ code

```python
import cppyy

cppyy.cppdef('int add(int a, int b) { return a + b; }')
cppyy.gbl.add(2, 3)
```

`cppdef()` parses and declares C++ code at runtime. `cppyy.gbl` exposes the global namespace. cppyy relies on Cling and a compatible C++ toolchain; installation support varies by platform.

## Define and instantiate classes

```python
cppyy.cppdef('''
class Counter { public: int value = 0; void increment() { ++value; } };
''')
counter = cppyy.gbl.Counter()
counter.increment(); counter.value
```

C++ classes declared through cppyy can be instantiated and used like Python objects. Public fields and methods are exposed through attribute access. C++ access rules still apply.

## Headers and standard library

```python
cppyy.include('vector')
values = cppyy.gbl.std.vector[int]()
values.push_back(4); values.push_back(7)
list(values)
```

`include()` loads a C++ header. Standard library templates can be instantiated with Python subscription syntax. Supported C++ containers expose Python iteration protocols.

## Load libraries and namespaces

```python
# cppyy.include('project/header.hpp')
# cppyy.load_library('libproject')
cppyy.cppdef('namespace geometry { double square(double x) { return x*x; } }')
cppyy.gbl.geometry.square(3.0)
```

`include()` exposes declarations from headers; `load_library()` loads compiled implementations. Namespaces are accessible as nested attributes under `cppyy.gbl`. Header declarations and binary symbols must match.
