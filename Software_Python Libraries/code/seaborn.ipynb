# Seaborn

Intermediate reference for statistical visualization with tidy data and Matplotlib integration.

## Setup

```python
import seaborn as sns
import matplotlib.pyplot as plt

tips = sns.load_dataset('tips')
sns.set_theme(style='whitegrid', context='notebook')
```

`load_dataset()` loads Seaborn example data and may require an internet connection on first use. `set_theme()` configures visual defaults. Seaborn plots can be further customized using Matplotlib axes.

## Distributions

```python
sns.histplot(data=tips, x='total_bill', hue='sex', kde=True)
sns.kdeplot(data=tips, x='total_bill', fill=True)
sns.ecdfplot(data=tips, x='total_bill')
```

`histplot()` displays binned frequencies or densities. `kdeplot()` estimates a smooth distribution; `ecdfplot()` shows the empirical cumulative distribution. `rugplot()` marks individual observations. Histogram bins and KDE bandwidth affect interpretation.

## Relationships

```python
sns.scatterplot(data=tips, x='total_bill', y='tip', hue='day', size='size')
sns.lineplot(data=tips, x='size', y='tip')
```

`scatterplot()` maps variables to point position, color, and size. `lineplot()` estimates trends across an ordered variable and can display uncertainty intervals. `relplot()` provides a figure-level interface with facets.

## Categorical plots

```python
sns.boxplot(data=tips, x='day', y='total_bill')
sns.violinplot(data=tips, x='day', y='total_bill')
sns.countplot(data=tips, x='day')
```

`boxplot()` compares medians, quartiles, and potential outliers. `violinplot()` adds a distribution estimate. `countplot()` counts observations; `barplot()` estimates a statistic; `stripplot()` and `swarmplot()` show individual points.

## Correlation and facets

```python
sns.heatmap(tips.select_dtypes('number').corr(), annot=True, center=0)
sns.pairplot(tips, vars=['total_bill','tip','size'], hue='sex')
```

`heatmap()` encodes a matrix using color; `DataFrame.corr()` calculates pairwise correlations. `pairplot()` compares pairwise variable relationships. Figure-level functions include `catplot()`, `displot()`, and `lmplot()`. Correlation does not establish causation.

## Regression visualization

```python
sns.regplot(data=tips, x='total_bill', y='tip', ci=95)
sns.lmplot(data=tips, x='total_bill', y='tip', hue='sex')
```

`regplot()` draws a fitted regression line on an axes. `lmplot()` provides a figure-level interface with grouping and facets. These are visualization tools; use a statistical model when formal inference and assumption checks are required.
