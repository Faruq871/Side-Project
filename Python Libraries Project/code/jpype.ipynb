# JPype

Intermediate reference for starting a JVM and using Java classes from Python. A compatible Java installation is required.

## Start JVM

```python
import jpype

if not jpype.isJVMStarted():
    jpype.startJVM()
```

`startJVM()` starts the Java Virtual Machine inside the Python process. Install a compatible Java runtime or JDK. The JVM generally starts only once per process; configure classpaths and JVM options at startup.

## Access Java classes

```python
ArrayList = jpype.JClass('java.util.ArrayList')
items = ArrayList()
items.add('alpha'); items.add('beta')
items.size(), items.get(0)
```

`JClass()` loads a Java class by fully qualified name. Java methods are called through the returned class or instance. The class must be available on the JVM classpath.

## Arrays and primitive types

```python
JInt = jpype.JInt
values = jpype.JArray(JInt)([1,2,3])
values[0], len(values)
```

`JArray()` creates Java arrays with a specified component type. JPype provides primitive wrappers such as `JInt`, `JDouble`, and `JBoolean`. Explicit types help resolve overloaded Java method signatures.

## Packages and collections

```python
from jpype import JPackage, JClass

System = JPackage('java').lang.System
System.getProperty('java.version')
HashMap = JClass('java.util.HashMap')
mapping = HashMap(); mapping.put('a', 10)
```

`JPackage()` provides dynamic access to Java namespaces. `JClass()` loads a specific class. Java collections can often be iterated or converted to Python containers; custom Java objects may require explicit handling.
