# Pandas

Intermediate reference for importing, inspecting, cleaning, transforming, grouping, and reshaping tabular data.

## Create and inspect

```python
import pandas as pd

df = pd.DataFrame({'team':['A','B','A'], 'score':[8,5,9]})
df.head(), df.shape, df.dtypes, df.describe(include='all')
```

`DataFrame()` creates a table from dictionaries or arrays. `head()` previews rows; `shape` reports dimensions; `dtypes` shows column types; `describe()` summarizes columns. Use `info()` for non-null counts and memory usage.

## Read and write files

```python
df = pd.read_csv('data.csv')
df.to_csv('clean.csv', index=False)
```

`read_csv()` reads delimited text. The `pd.read_*()` family also includes `read_excel()`, `read_json()`, `read_parquet()`, `read_sql()`, `read_html()`, `read_xml()`, and `read_feather()`. Writers include `to_excel()`, `to_json()`, `to_parquet()`, and `to_sql()`. Some formats require optional engines.

## Select data

```python
df['score'], df[['team','score']]
df.loc[df.score > 6, ['team','score']]
df.iloc[:2, :]
```

`[]` selects columns or filters rows. `loc` selects by labels and conditions; `iloc` selects by integer position. `.at` and `.iat` access individual cells. Prefer `.loc` for explicit assignment.

## Clean and transform

```python
df.isna().sum()
df.dropna(subset=['score']).drop_duplicates()
df.assign(score_scaled=lambda x: x.score / 10)
```

`isna()` and `notna()` identify missing values. `dropna()` removes missing rows or columns; `fillna()` replaces missing values. `drop_duplicates()` removes repeated rows. `assign()` creates or replaces columns. String operations are available through `.str`, such as `strip()`, `lower()`, `contains()`, and `replace()`.

## Group and combine

```python
df.groupby('team').score.agg(['count','mean','max'])
left = pd.DataFrame({'id':[1,2], 'x':[10,20]})
right = pd.DataFrame({'id':[1,2], 'y':[3,4]})
left.merge(right, on='id', how='left')
```

`groupby()` splits rows into groups; `agg()` applies one or more summaries. `transform()` returns group calculations aligned to original rows. `merge()` performs database-style joins (`inner`, `left`, `right`, `outer`); `concat()` stacks objects by rows or columns.

## Reshape and dates

```python
df.pivot_table(index='team', values='score', aggfunc='mean')
df.melt(id_vars='team', value_vars=['score'])
pd.to_datetime(['2025-01-01','2025-01-02'])
```

`pivot_table()` reshapes and aggregates; `pivot()` reshapes without aggregation and requires unique index/column pairs. `melt()` converts wide data to long form. Time-series tools include `to_datetime()`, `resample()`, `rolling()`, `shift()`, and `DatetimeIndex`.
