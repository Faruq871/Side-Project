# Plotly

Intermediate reference for interactive charts with Plotly Express and graph objects.

## Plotly Express

```python
import plotly.express as px

df = px.data.iris()
fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species', size='petal_length')
fig.show()
```

Plotly Express provides concise functions such as `scatter()`, `line()`, `bar()`, `histogram()`, `box()`, `violin()`, and `imshow()`. `show()` renders the interactive figure in a notebook or browser.

## Customize charts

```python
fig.update_layout(title='Iris measurements', template='plotly_white', xaxis_title='Sepal width')
fig.update_traces(marker={'opacity':.7})
```

`update_layout()` changes titles, dimensions, legend placement, and templates. `update_traces()` modifies markers, lines, and hover behavior. `update_xaxes()` and `update_yaxes()` customize axis ranges and labels.

## Graph objects

```python
import plotly.graph_objects as go

fig = go.Figure()
fig.add_trace(go.Scatter(x=[1,2,3], y=[2,1,4], mode='lines+markers', name='Series'))
fig.add_trace(go.Bar(x=['A','B'], y=[3,5], name='Counts'))
```

`graph_objects` provides explicit control over chart traces and layout. `Figure()` creates a figure and `add_trace()` adds layers. Trace types include `Scatter`, `Bar`, `Histogram`, `Heatmap`, `Box`, and `Violin`.

## Facets and subplots

```python
fig = px.scatter(df, x='sepal_width', y='sepal_length', color='species', facet_col='species')
from plotly.subplots import make_subplots
sub = make_subplots(rows=1, cols=2)
sub.add_trace(go.Scatter(x=[1,2], y=[2,3]), row=1, col=1)
```

`facet_row` and `facet_col` create small multiples. `make_subplots()` creates a grid of plotting areas; add traces with explicit row and column positions. Express supports `hover_data`, `symbol`, `animation_frame`, and `category_orders`.

## Export

```python
fig.write_html('chart.html')
fig.write_image('chart.png')
```

`write_html()` exports an interactive HTML file. Static export through `write_image()` requires Kaleido. Other supported formats include SVG and PDF.
