# Py4J

Intermediate reference for Python-to-Java communication through a gateway. A Java gateway server must be running.

## Connect to gateway

```python
from py4j.java_gateway import JavaGateway

gateway = JavaGateway()
jvm = gateway.jvm
```

`JavaGateway()` connects to a running Py4J gateway server, commonly started by a Java application. Defaults use localhost and the standard gateway port. Configure address, port, authentication, and TLS for other environments.

## Access Java classes

```python
ArrayList = jvm.java.util.ArrayList
items = ArrayList()
items.add('one'); items.add('two')
items.size()
```

`gateway.jvm` exposes Java packages and classes through a dynamic proxy. Java objects are represented by Python proxy objects. Each call crosses a process boundary, so batch operations when possible.

## Call Java entry point

```python
entry_point = gateway.entry_point
# result = entry_point.calculate(10, 20)
```

`gateway.entry_point` refers to the Java object registered as the gateway's entry point. Define methods in the Java application for Python callers. The commented call is illustrative; its method must exist on the running entry point.

## Convert collections and close

```python
from py4j.java_collections import ListConverter, MapConverter

java_list = ListConverter().convert([1,2,3], gateway._gateway_client)
java_map = MapConverter().convert({'x':1}, gateway._gateway_client)
gateway.close()
```

`ListConverter` and `MapConverter` convert Python collections into Java collection proxies. `close()` closes the gateway connection. If Python owns the Java process, manage that process separately and handle network failures.
