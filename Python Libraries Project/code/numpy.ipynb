# NumPy

Intermediate reference for arrays, numerical operations, and linear algebra.

## Create arrays

```python
import numpy as np

a = np.array([[1, 2, 3], [4, 5, 6]])
np.zeros((2, 3)), np.arange(0, 10, 2), np.linspace(0, 1, 5)
```

`array()` converts sequences to arrays. `zeros()`, `ones()`, `full()`, and `empty()` create arrays with specified contents. `arange()` uses a step and excludes the stop value; `linspace()` creates a fixed number of evenly spaced values.

## Inspect and reshape

```python
a.shape, a.ndim, a.size, a.dtype
a.astype(np.float32).reshape(3, 2)
a.ravel(), a.flatten()
```

`shape`, `ndim`, `size`, and `dtype` describe dimensions, rank, element count, and element type. `astype()` converts types. `reshape()` changes dimensions; `ravel()` flattens with a view when possible, while `flatten()` returns a copy.

## Index and filter

```python
a[0, 1], a[:, 1], a[1, :]
a[a > 3]
a[(a >= 2) & (a <= 5)]
```

Use integer indices and slices to select elements, rows, or columns. Boolean masks filter values. Combine element-wise conditions with `&`, `|`, and `~`, with each comparison in parentheses. `where()` selects values conditionally.

## Broadcast and aggregate

```python
x = np.array([1, 2, 3])
np.ones((2, 3)) + x
np.mean(a, axis=0), np.sum(a, axis=1)
```

Broadcasting applies operations across compatible shapes without manually repeating data. Dimensions are compatible when equal or one is 1. Aggregation with `axis=0` reduces rows; `axis=1` reduces columns. Related functions include `min()`, `max()`, `std()`, `var()`, and `percentile()`.

## Missing values and sorting

```python
v = np.array([3., np.nan, 1., 3.])
np.nanmean(v), np.isnan(v)
np.sort(v[~np.isnan(v)]), np.unique(v)
```

Use `isnan()` and `isfinite()` to detect invalid numeric values. `nanmean()`, `nanmedian()`, and other `nan*` functions ignore NaNs. `sort()` orders values, `argsort()` returns sorting indices, and `unique()` returns distinct values.

## Random numbers and linear algebra

```python
rng = np.random.default_rng(42)
rng.normal(0, 1, 4), rng.integers(0, 10, 4)
A = np.array([[2., 1.], [1., 3.]])
np.linalg.solve(A, [1., 2.])
```

Use `default_rng()` to create a reproducible random generator; it provides `normal()`, `integers()`, `choice()`, and `permutation()`. `linalg.solve()` solves linear systems. Other routines include `det()`, `eig()`, `svd()`, `norm()`, and `lstsq()`. Prefer `solve()` over explicitly calculating an inverse.
