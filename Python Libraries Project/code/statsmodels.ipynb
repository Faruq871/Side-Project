# Statsmodels

Intermediate reference for regression, inference, diagnostics, and time-series models.

## Fit OLS

```python
import numpy as np
import pandas as pd
import statsmodels.api as sm

df = pd.DataFrame({'x':np.arange(1,11), 'y':2+3*np.arange(1,11)+np.random.default_rng(1).normal(size=10)})
fit = sm.OLS(df.y, sm.add_constant(df.x)).fit()
```

`OLS()` estimates ordinary least squares. `add_constant()` adds an intercept in the array-based API. `.fit()` returns a results object containing estimates, uncertainty, tests, and fit statistics.

## Inspect and formulate

```python
fit.summary()
fit.params, fit.bse, fit.pvalues, fit.conf_int()
import statsmodels.formula.api as smf
smf.ols('y ~ x', data=df).fit()
```

`summary()` reports coefficients, standard errors, test statistics, and diagnostics. `params`, `bse`, `pvalues`, and `conf_int()` expose individual results. The formula API uses `outcome ~ predictors`; `C(variable)` marks categorical predictors and `*` adds interactions.

## Hypothesis tests and ANOVA

```python
from statsmodels.stats.anova import anova_lm

anova_lm(fit, typ=2)
fit.t_test('x = 0')
fit.f_test('x = 0')
```

`anova_lm()` creates an ANOVA table for supported models. `t_test()` evaluates linear coefficient restrictions; `f_test()` tests joint restrictions. Type I, II, and III sums of squares answer different questions, especially with unbalanced designs.

## Diagnostics and robust errors

```python
from statsmodels.stats.diagnostic import het_breuschpagan, acorr_ljungbox
from statsmodels.stats.stattools import durbin_watson

durbin_watson(fit.resid)
het_breuschpagan(fit.resid, fit.model.exog)
acorr_ljungbox(fit.resid, lags=[1,5], return_df=True)
smf.ols('y ~ x', data=df).fit(cov_type='HC3').bse
```

These functions test residual autocorrelation and heteroskedasticity. `cov_type='HC3'` requests heteroskedasticity-robust standard errors. Other covariance options include HAC and clustered estimators. Robust errors do not fix omitted variables, endogeneity, or misspecified models.

## Time series

```python
from statsmodels.tsa.arima.model import ARIMA

series = pd.Series([10,12,11,13,15,14,16,18])
ts_fit = ARIMA(series, order=(1,1,1)).fit()
ts_fit.forecast(steps=3)
```

`ARIMA()` fits autoregressive integrated moving-average models with order `(p,d,q)`. `forecast()` produces out-of-sample predictions. Inspect stationarity and residuals, and evaluate forecasts with time-ordered validation.
