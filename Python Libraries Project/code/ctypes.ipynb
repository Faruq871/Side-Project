# ctypes

Intermediate reference for calling C functions in shared libraries. ctypes is part of the Python standard library.

## Load shared library

```python
import ctypes
from ctypes.util import find_library

name = find_library('c')
libc = ctypes.CDLL(name) if name else None
```

`CDLL()` loads a library using the standard C calling convention. `WinDLL()` is used for Windows stdcall. `find_library()` may locate a system library, but names and paths vary by operating system.

## Declare function signatures

```python
if libc:
    libc.strlen.argtypes = [ctypes.c_char_p]
    libc.strlen.restype = ctypes.c_size_t
    libc.strlen(b'python')
```

Set `argtypes` and `restype` for every called function. These declarations control argument conversion and return interpretation. Incorrect signatures can corrupt memory or crash the process; verify them against C headers.

## Pointers and primitive types

```python
value = ctypes.c_int(42)
pointer = ctypes.pointer(value)
pointer.contents.value
```

ctypes provides C-compatible types such as `c_int`, `c_double`, `c_char_p`, and `c_size_t`. `pointer()` creates a pointer; `byref()` is often used for output parameters. Keep referenced objects alive while native code uses their memory.

## Structures and callbacks

```python
class Point(ctypes.Structure):
    _fields_ = [('x', ctypes.c_double), ('y', ctypes.c_double)]
p = Point(1.5, 2.5)
CALLBACK = ctypes.CFUNCTYPE(ctypes.c_int, ctypes.c_int)
callback = CALLBACK(lambda x: x * 2)
```

`Structure` maps a C struct; `_fields_` must match field order and layout. `CFUNCTYPE()` creates a C-callable callback; `WINFUNCTYPE()` is used for stdcall on Windows. Keep callback objects alive while native code may call them.
