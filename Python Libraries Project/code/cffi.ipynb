# CFFI

Intermediate reference for declaring C interfaces and calling native libraries.

## Declare C interface

```python
from cffi import FFI

ffi = FFI()
ffi.cdef('int abs(int);')
```

`FFI()` creates an interface object. `cdef()` declares C types and signatures. Declarations must match the actual library ABI; use C headers as the source of truth.

## Load and call library

```python
import ctypes.util
name = ctypes.util.find_library('c')
C = ffi.dlopen(name) if name else None
if C:
    C.abs(-12)
```

`dlopen()` loads a shared library and exposes declared symbols. The library must export the declared functions. For production, use a known configurable library path rather than relying only on system lookup.

## Allocate C data

```python
number = ffi.new('int *', 10)
values = ffi.new('int[]', [1,2,3])
number[0], values[1]
```

`new()` allocates C data using a declared type. Pointer types use `*`; arrays use `[]`. CFFI manages allocated memory while references remain alive. Inspect types with `typeof()` and sizes with `sizeof()`.

## Convert buffers and strings

```python
text = ffi.new('char[]', b'hello')
ffi.string(text)
ffi.buffer(values, ffi.sizeof('int') * 3)
```

`string()` copies a NUL-terminated C string into Python bytes. `buffer()` exposes C memory through Python's buffer protocol. `cast()` reinterprets values as C types and does not guarantee semantic safety.

## ABI and API modes

```python
# ABI mode: ffi.dlopen() loads an existing shared library.
# API mode: ffi.set_source() compiles a C extension against headers.
```

ABI mode relies on declarations and dynamic loading. API mode compiles an extension using C headers and offers compiler checking. API mode requires a C compiler and development headers.
