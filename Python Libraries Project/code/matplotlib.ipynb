# Matplotlib

Intermediate reference for static plots using Matplotlib's object-oriented interface.

## Create plots

```python
import matplotlib.pyplot as plt

fig, ax = plt.subplots(figsize=(7,4))
ax.plot([1,2,3], [2,4,3], marker='o')
ax.set(title='Example', xlabel='X', ylabel='Y')
plt.show()
```

`subplots()` creates a figure and axes. Prefer methods on `ax` for reusable plotting code. Common methods include `plot()`, `scatter()`, `bar()`, `hist()`, `boxplot()`, `imshow()`, and `errorbar()`.

## Multiple chart types

```python
fig, axes = plt.subplots(2, 2, figsize=(8,6))
axes[0,0].scatter([1,2,3],[3,1,4])
axes[0,1].bar(['A','B'],[4,7])
axes[1,0].hist([1,1,2,3,3,3])
axes[1,1].boxplot([[1,2,3],[2,4,8]])
```

An axes object represents one plotting area. Use indexing to access axes in a grid. Choose a chart type based on the data: scatter for paired observations, histogram for distributions, and boxplots for spread and potential outliers.

## Format axes and legends

```python
ax.set_xlim(0,10); ax.set_ylim(0,20)
ax.set_xticks([0,5,10]); ax.grid(alpha=.3)
ax.plot([1,2,3],[2,3,5], label='Observed')
ax.legend()
```

`set_xlim()` and `set_ylim()` control ranges. `set_xticks()` and `set_yticks()` set tick positions. `grid()` adds grid lines; `legend()` displays labels supplied to plotted objects. Use `tick_params()` to format tick marks and labels.

## Annotations and layout

```python
ax.annotate('Peak', xy=(3,5), xytext=(2,5.5), arrowprops={'arrowstyle':'->'})
fig.tight_layout()
fig.savefig('figure.png', dpi=300, bbox_inches='tight')
```

`annotate()` places explanatory text and can point to a data coordinate. `tight_layout()` reduces overlap. `savefig()` exports PNG, PDF, SVG, and other formats; `dpi` controls raster resolution. Use `constrained_layout=True` in `subplots()` as another layout option.

## Styles and configuration

```python
plt.style.use('ggplot')
plt.rcParams.update({'font.size':11, 'figure.dpi':120})
```

`style.use()` applies a built-in or custom style. `rcParams` controls defaults such as font size and figure resolution. Inspect available styles with `plt.style.available`; reset defaults with `plt.rcdefaults()`.
